```python
from google.colab import drive #Permite usar los archivos del usuario.
import pandas as pd # Librería que permite la manipulación y análisis de datos.
import plotly.graph_objects as go # Librería que permite la creación de gráficas interactivas.
```

```python
drive.mount('/content/drive', force_remount=True) # Monta el disco de la nube del usuario. Si ya se encuentra montada, vuelve a hacerlo.
wd = "/content/drive/MyDrive/Forecast_FIME/PIA/"
```

```python
df_original = pd.read_csv(wd + "df_USDMXN_20250330.csv")
df_original.rename(columns = {"Unnamed: 0": "fecha"}, inplace = True)
df_original.set_index("fecha", inplace = True)

df_arima = pd.read_csv(wd + "forecast_USDMXN_ARIMA.csv")
df_arima.rename(columns = {"Unnamed: 0": "fecha", "predicted_mean": "dato"}, inplace = True)
df_arima.set_index("fecha", inplace = True)

df_holtwinter_auto = pd.read_csv(wd + "forecast_USDMXN_HoltWinter_auto.csv")
df_holtwinter_auto.rename(columns = {"Unnamed: 0": "fecha", "0": "dato"}, inplace = True)
df_holtwinter_auto.set_index("fecha", inplace = True)

df_holtwinter_manual = pd.read_csv(wd + "forecast_USDMXN_HoltWinter_manual.csv")
df_holtwinter_manual.rename(columns = {"Unnamed: 0": "fecha", "0": "dato"}, inplace = True)
df_holtwinter_manual.set_index("fecha", inplace = True)

df_prophet = pd.read_csv(wd + "forecast_USDMXN_Prophet.csv")
df_prophet.rename(columns = {"ds": "fecha", "yhat": "dato"}, inplace = True)
df_prophet.set_index("fecha", inplace = True)
```

```python
df_holtwinter_auto
```

```python
fig = go.Figure()

## ARIMA
fig.add_trace(
    go.Scatter(
        x=df_arima.index,
        y=df_arima["dato"],
        name = "ARIMA",
          line = dict(color = "#ff7f0e", width = 3, dash = "dash")
    )
)

## Holt-Winter (auto)
fig.add_trace(
    go.Scatter(
        x=df_holtwinter_auto.index,
        y=df_holtwinter_auto["dato"],
        name = "Holt-Winter (auto)",
          line = dict(color = "#2ca02c", width = 3, dash = "dash")
    )
)

## Holt-Winter (manual)
fig.add_trace(
    go.Scatter(
        x=df_holtwinter_manual.index,
        y=df_holtwinter_manual["dato"],
        name = "Holt-Winter (manual)",
          line = dict(color = "#d62728", width = 3, dash = "dash")
    )
)

## Prophet
fig.add_trace(
    go.Scatter(
        x=df_prophet.index,
        y=df_prophet["dato"],
        name = "Prophet",
          line = dict(color = "#9467bd", width = 3, dash = "dash")
    )
)

## Original
fig.add_trace(
    go.Scatter(
        x=df_original.index,
        y=df_original["dato"],
        name = "Real",
          line = dict(color = "#1f77b4", width = 3, dash = "dash")
    )
)

fig.update_layout(title = "Tasa de cambio real vs. pronosticada (comparativa)", xaxis_title="Tiempo", yaxis_title="Tasa de cambio")

fig.show()
```
